```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv('housing_dataset.csv')
```

```python
data['age'] = 2015 - data['yr_built']
data['last_updated'] = 2015 - data[['yr_built','yr_renovated']].max(axis=1)
data = data.drop(columns = ['yr_built','yr_renovated'])
```

```python
data = data.drop(columns=['zipcode','date','id'])
```

```python
data.head()
```

```python
X = np.array(data.iloc[:,1:])
y = np.array(data.iloc[:,0])
```

```python
print(f"Shape of X: {X.shape}")
print(f"Shape of y: {y.shape}")
```

```python
"""
  Step 1: Divide the data into splits (Test, Train, Dev)
  Step 2: Normalize the data
  Step 3: Train the model (improve the loss)
  Step 4: Predictions and Metrics
"""
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=59)
```

```python
print(X_train.shape,X_test.shape)
```

```python
print(y_train.shape,y_test.shape)
```

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

X_train = (X_train - mean) / std
X_test  = (X_test  - mean) / std
```

```python
y_mean = y_train.mean(axis=0)
y_std =  y_train.std(axis=0)

y_train = (y_train - y_mean) / y_std
y_test  = (y_test  - y_mean) / y_std
```

```python
def loss(y_actual,y_pred):
  m = y_actual.shape[0]
  error_square = (y_pred - y_actual) ** 2
  total_error = np.sum(error_square)
  loss = total_error / ( 2 * m)
  return loss
```

```python
def train_model(num_iters,X,y,w,b,alpha):
  losses = []
  m = X.shape[0]

  for i in range(num_iters):
    if i>=10000:
      alpha = 0.00001
    y_pred = np.dot(X,w)+b

    #params update and loss

    current_loss = loss(y,y_pred)
    losses.append(current_loss)

    error = (y_pred - y) / m

    dw = np.dot(X.T,error)
    db = np.sum(error)

    w = w - alpha * (dw)
    b = b - alpha * (db)

    if i % 1000 == 0:
      print(f"Iteration {i}: Loss {current_loss}")

  return w,b,losses

```

```python
w = np.random.randn(X_train.shape[1],1)
b = 0
alpha = 0.01 #learning_parameter
y_train = y_train.reshape(-1,1)
y_test = y_test.reshape(-1,1)
w, b, loss_history = train_model(3000, X_train, y_train, w, b, alpha)
```

```python
def predict(X,w,b):
  y_pred = np.dot(X,w)+b
  return y_pred
```

```python
predictions = predict(X_test,w,b)
print(loss(y_test,y_pred=predictions))
```
